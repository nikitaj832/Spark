```python
df = spark.read.format("csv").option("header","True").load("/Volumes/abc/upload_d/a_volume/sales_data.csv",header=True,inferSchema =True)
```

```python
from pyspark.sql.functions import col, sum, when, trim, split, regexp_extract, year, month, to_timestamp

missing_values = df.select([
    sum(when(col(c).isNull() | (trim(col(c)) == "null"), 1).otherwise(0)).alias(c)
    for c in df.columns
])
missing_values.show()
```

```python
df_clean = df.filter(
    (col("Order ID").isNotNull()) & (trim(col("Order ID")) != "null") &
    (col("Product").isNotNull()) & (trim(col("Product")) != "null") &
    (col("Quantity Ordered").isNotNull()) & (trim(col("Quantity Ordered")) != "null") &
    (col("Price Each").isNotNull()) & (trim(col("Price Each")) != "null") &
    (col("Order Date").isNotNull()) & (trim(col("Order Date")) != "null") &
    (col("Purchase Address").isNotNull()) & (trim(col("Purchase Address")) != "")
)
```

```python
df_clean.show(5)
```

```python
df_clean.describe().show()
```

```python
df_extracted = df_clean.withColumn("Pincode", regexp_extract(col("Purchase Address"), r"(\d{5})$", 1)) \
                       .withColumn("City", regexp_extract(col("Purchase Address"), r",\s*(\w[\w\s]+),", 1))
df_extracted.display()
```

```python
df_extracted.describe().show()
```

```python
from pyspark.sql.functions import to_timestamp
from pyspark.sql.types import IntegerType, FloatType
```

```python
from pyspark.sql.functions import desc

top_5_cities = df_extracted.groupBy("City") \
                       .count() \
                       .orderBy(desc("count")) \
                       .limit(5)

top_5_cities.display()
```

```python
from pyspark.sql.functions import desc

top_3_pincodes = df_extracted.groupBy("Pincode") \
                         .count() \
                         .orderBy(desc("count")) \
                         .limit(3)

top_3_pincodes.show()
```

```python
from pyspark.sql.functions import to_timestamp, col, year, sum as sum_

# Step 1: Filter out rows with bad 'Order Date' values
df_valid = df_extracted.filter(
    (col("Order Date").isNotNull()) & 
    (col("Order Date") != "Order Date") & 
    (col("Order Date") != "null")
)

# Step 2: Convert 'Order Date' to timestamp
df_with_date = df_valid.withColumn(
    "Order_Timestamp", 
    to_timestamp(col("Order Date"), "MM/dd/yy HH:mm")\

    )
```

```python
# Step 3: Cast quantity and price to double
df_with_revenue = df_with_date.withColumn("Quantity", col("Quantity Ordered").cast("double")) \
    .withColumn("Price", col("Price Each").cast("double")) \
    .withColumn("year", year(col("Order_Timestamp"))) \
    .withColumn("month", month(col("Order_Timestamp"))) \
    .withColumn("revenue", col("Quantity") * col("Price"))
```

```python


df_with_revenue.display()
```

```python

# Step 4: Group by year and sum revenue
df_with_revenue.groupBy("year") \
    .agg(sum_("revenue").alias("total_revenue")) \
    .orderBy("year") \
    .display()
```

```python
df_with_revenue.describe().display()
```

```python
from pyspark.sql.functions import count

duplicate_orders = df_with_revenue.groupBy("Order ID") \
                     .agg(count("*").alias("count")) \
                     .filter("count > 1")

display(duplicate_orders)
```

```python
from pyspark.sql.functions import collect_list

# Group products per Order ID into a list (includes duplicates)
order_products = df_with_revenue.groupBy("Order ID") \
                   .agg(collect_list("Product").alias("products"))



order_products.display()
```

```python
order_products = order_products.filter("size(products) > 1")
```

```python
from pyspark.sql.types import ArrayType, StringType
from pyspark.sql.functions import udf
from itertools import combinations

def get_combinations(products):
    return [' & '.join(sorted(pair)) for pair in combinations(products, 2)]

combo_udf = udf(get_combinations, ArrayType(StringType()))

# Apply UDF
combinations_df = order_products.withColumn("product_combos", combo_udf("products"))

```

```python
from pyspark.sql.functions import explode, desc

# Flatten and count
combo_counts = combinations_df.select(explode("product_combos").alias("combo")) \
                              .groupBy("combo") \
                              .count()  \
                              .orderBy(desc("count"))
combo_counts.display()
```

```python
top3_combos = combo_counts.limit(3)
display(top3_combos)
```
